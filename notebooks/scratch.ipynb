```python
rng = np.random.default_rng(seed=1)
n = 3000

dff = pd.DataFrame({
    "age": rng.integers(18, 80, size=n),
    "region": rng.choice(["Urban", "Suburban", "Rural"], size=n, p=[0.4, 0.35, 0.25]),
    "smoker": rng.choice(["no", "yes"], size=n, p=[0.8, 0.2]),
})
```

```python
dff
```

```python
log_mean = 7.6 + 0.01 * (dff["age"] - 40) + 0.35 * (dff["smoker"] == "yes")
dff["cost"] = rng.lognormal(mean=log_mean, sigma=0.8)

dff.head()
```

```python
fig, ax = plt.subplots()
ax.hist(dff["cost"], bins=40, color="steelblue", edgecolor="white")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].hist(np.log10(dff["cost"]), bins = 30, color="steelblue", edgecolor="white")
axes[0].set_title("Cost distribution (log10)")
axes[0].set_xlabel("log10 of claim cost")

axes[1].scatter(dff["age"], dff["cost"], alpha=0.2, s = 8)
axes[1].set_yscale("log")
axes[1].set_title("Cost vs age")
axes[1].set_xlabel("Age")


fig.suptitle("Claim cost overview")
fig.tight_layout()
plt.show()

```

```python
mean_by_age = df.groupby(pd.cut(df["age"], bins=[17, 30, 45, 60, 80]), observed=True)["charges"].mean()
print(mean_by_age)
labels = [str(interval) for interval in mean_by_age.index]

fig, ax = plt.subplots(figsize=(8, 6))
bars = ax.bar(labels, mean_by_age.values, color="#4C78A8")

overall = df["charges"].mean()
ax.axhline(overall, color="crimson", linestyle="--", label=f"Overall mean: {overall:,.0f}")

ax.bar_label(bars, fmt="{:,.0f}", padding=3)
ax.set_xlabel("Age group")
ax.set_ylabel("Average claim cost")
ax.set_title("Average claim cost by age group")
ax.grid(axis="y", alpha=0.3)
ax.legend()
fig.tight_layout()
plt.show()
```
