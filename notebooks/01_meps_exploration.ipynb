```python
spend = df.loc[df["TOTEXP23"] > 0, "TOTEXP23"]
num_people = len(df)
num_spend = len(spend)
num_zero = num_people - num_spend

print("Total number of people:", num_people)
print("Number of people who spend:", num_spend, f"({num_spend/num_people*100:.2f}%)")
print("Number of people who don't spend:", num_zero, f"({num_zero/num_people*100:.2f}%)" )

print(spend.describe())
q = spend.quantile([.9, .95, .99])
print(q)
print(q.iloc[0])
top_10_percent = df[df["TOTEXP23"] > q.iloc[0]]["TOTEXP23"]
print(top_10_percent.sum())
print(df["TOTEXP23"].sum())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15,5))
axes[0].hist(spend, bins=50, color="steelblue", edgecolor="white")
axes[0].set_title("Dollars")
axes[0].set_xlabel("Total expenditure, 2023 ($)")
axes[0].set_ylabel("Number of people")


axes[1].hist(np.log10(spend), bins=50, color="steelblue", edgecolor="white")
axes[1].set_title("Log scale")
axes[1].set_xticks([1,2, 3, 4, 5])
axes[1].set_xticklabels(["$10","$100", "$1K", "$10K", "$100K"])
axes[1].set_xlabel("Total expenditure, 2023 (log scale)")
axes[1].set_ylabel("Number of people")



fig.suptitle(
    "Distribution of health care spending among people with any spending, MEPS 2023", 
    fontsize=14,
)
fig.tight_layout()
plt.show()
```

What the plots show:
- Left (dollars): almost everyone is in the first bar, about 13,000 of the roughly 16,300 people. This axis runs out to $600,000 for a few extreme spenders you can't even see.
- Right (log scale): a smooth, nearly symmetric hump. The peak is around 3.5 to 3.7, which is roughly $3,000 to $5,000. The range runs from about 1 to about 5.7, so spending goes from a few dollars to around half a million. There's slightly longer tail on the left, which matches the mild negative skew I measured earlier.

What this mean:
- A log-based model fits: Spending among spenders is roughly lognormal, so a GLM with a log link, or a model of log(spend), matches the data's shape. 
- The mean is misleading here. The left panel shows why: a few huge values pull the average up, so the median is the better "typical person" number. 
- The zeros are a separate group.
